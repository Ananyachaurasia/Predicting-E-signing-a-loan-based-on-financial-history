### FEATURE ENGINEERING

```python
df=df.drop(columns=["months_employed"])
```

```python
df.head()
```

```python
df["personal_account_months"]=(df.personal_account_m+(df.personal_account_y*12))
```

```python
df.head()
```

```python
df=df.drop(columns=["personal_account_m","personal_account_y"])
```

```python
df.head()
```

### DATA PREPROCESSING

```python
df=pd.get_dummies(df)
```

```python
df.columns
```

```python
df=df.drop(columns=["pay_schedule_semi-monthly"])
```

```python
y=df["e_signed"]
```

```python
users=df["Entry_id"]
```

```python
df=df.drop(columns=["Entry_id","e_signed"])
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(df,y,test_size=0.2,random_state=42)
```

```python
df.head()
```

```python
from sklearn.preprocessing import StandardScaler
scalar=StandardScaler()
X_train=scalar.fit_transform(X_train)
X_test=scalar.transform(X_test)
```

```python
X_train
```

```python
y_train
```

### MODEL BUILDING

```python
from sklearn.linear_model import LogisticRegression
classifer=LogisticRegression(random_state=0,penalty="l1",solver='liblinear')
classifer.fit(X_train,y_train)
y_pred=classifer.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,precision_score,recall_score,f1_score
acc=accuracy_score(y_test,y_pred)
pre=precision_score(y_test,y_pred)
rec=recall_score(y_test,y_pred)
f1=f1_score(y_test,y_pred)
```

```python
results=pd.DataFrame([["Logistic Regression",acc,pre,rec,f1]],columns=["model","accuracy","precison","recall","f1"])
```

```python
results
```

```python
from sklearn.svm import SVC
classifer=SVC(random_state=0,kernel="linear")
classifer.fit(X_train,y_train)
y_pred=classifer.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,precision_score,recall_score,f1_score
acc=accuracy_score(y_test,y_pred)
pre=precision_score(y_test,y_pred)
rec=recall_score(y_test,y_pred)
f1=f1_score(y_test,y_pred)
```

```python
model_results=pd.DataFrame([["SVM(linear)",acc,pre,rec,f1]],columns=["model","accuracy","precison","recall","f1"])
```

```python
model_results
```

```python
from sklearn.svm import SVC
classifer=SVC(random_state=0,kernel="rbf")
classifer.fit(X_train,y_train)
y_pred=classifer.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,precision_score,recall_score,f1_score
acc=accuracy_score(y_test,y_pred)
pre=precision_score(y_test,y_pred)
rec=recall_score(y_test,y_pred)
f1=f1_score(y_test,y_pred)
```

```python
model_results=pd.DataFrame([["SVM(rbf)",acc,pre,rec,f1]],columns=["model","accuracy","precison","recall","f1"])
```

```python
model_results
```

```python
from sklearn.ensemble import RandomForestClassifier
classifer=RandomForestClassifier(random_state=0,n_estimators=100,criterion="entropy")
classifer.fit(X_train,y_train)
y_pred=classifer.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,precision_score,recall_score,f1_score
acc=accuracy_score(y_test,y_pred)
pre=precision_score(y_test,y_pred)
rec=recall_score(y_test,y_pred)
f1=f1_score(y_test,y_pred)
```

```python
model_results=pd.DataFrame([["Random forest",acc,pre,rec,f1]],columns=["model","accuracy","precison","recall","f1"])
```

```python
model_results
```

### CROSS VALIDATION

```python
from sklearn.model_selection import cross_val_score
accuarcies=cross_val_score(classifer,X=X_train,y=y_train,cv=10)
print("Random forest : %0.2f(+/- %0.2f)" %(accuarcies.mean(),accuarcies.std()*2))
```
